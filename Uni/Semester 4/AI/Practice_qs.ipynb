```python
import heapq as heap
grid = [
    ['S', 1, 2, '#', 3],
    [1, '#', 2, '#', '#'],
    [1, 2, 1, 2, '#'],
    ['#', '#', 1, '#', '#'],
    [1, 1, 1, 2, 'T']
]

def get_neighbor(node):
    r,c = node
    directions = [(1,0), (-1,0), (0, 1), (0,-1)]
    neighbors = []

    for dr, dc in directions:
        nr = r + dr
        nc = c + dc
        if (0 <= nr and nr< 5) and (0 <= nc and nc < 5):
            val = grid[nr][nc]
            if val == '#':
                continue
            elif val == 'S' or val == 'T':
                cost = 0
            else:
                cost = val
            
            neighbors.append(((nr, nc), cost))
    return neighbors
def heuristic(node, goal):
    heur = abs(node[0]- goal[0]) + abs(node[1] - goal[1])
    return heur

def astar(start, goal, get_neighbors, heur):
    pq = []
    g = {start: 0}
    parent = {start: None}
    heap.heappush(pq, (g[start] + heur(start, goal), start))
    while pq:
        priority, node = heap.heappop(pq)
        if node == goal:
            path = []
            current = node
            while current is not None:
                path.append(current)
                current = parent[current]
            return goal, g[node], path[::-1]
        for n, c in get_neighbors(node):
            newg = g[node] + c
            if n not in g or newg < g[n]:
                g[n] = newg
                parent[n] = node
                heap.heappush(pq, (g[n] + heur(n, goal), n))
    return None, float('inf'), []

for i in range(5):
    for j in range(5):
        if grid[i][j] == 'S':
            start = (i,j)
        if grid[i][j] == 'T':
            goal = (i, j)
n, p, path= astar(start, goal, get_neighbor, heuristic)
print(f"Cost = {p}, and Node = {n}")
print(path)
```

```python
import random as rand

def fitness(x):
    return x*x + x

def create_population():
    population = []
    for _ in range(20):
        n = rand.randint(0, 31)
        binary = format(n, '05b')
        population.append(binary)
    return population

def crossover(parent1, parent2):
    child1 = parent1[0:3] + parent2[3:5]
    child2 = parent2[0:3] + parent1[3:5]
    return child1, child2

def mutation(child):
    index = rand.randint(0, 4)
    newchild = ''
    for i in range(5):
        if i == index:
            if child[i] == '1':
                newchild += '0'
            else:
                newchild += '1'
        else:
            newchild += child[i]

    return newchild
def selection(population):
    population_sorted = sorted(population, key = lambda x:fitness(int(x,2)), reverse= True)
    return population_sorted[0], population_sorted[1]
def genetic(population, fitness, crossover, mutate, generations):
    for i in range(generations):
        newpopulation = []
        while len(newpopulation) < len(population):
            parent1, parent2 = selection(population)
            c1, c2 = crossover(parent1, parent2)
            c1 = mutate(c1)
            c2 = mutate(c2)
            newpopulation.append(c1)
            if len(newpopulation) < len(population):
                newpopulation.append(c2)
        population = newpopulation
    return max(population, key= lambda x: fitness(int(x,2)))


population = create_population()
generations = 50
result = genetic(population, fitness, crossover, mutation, generations)
best_x = int(result, 2)
best_f = fitness(best_x)
print("Best chromosome:", result)
print("x =", best_x)
print("f(x) =", best_f)
```
